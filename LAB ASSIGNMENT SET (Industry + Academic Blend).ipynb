```python
# Build a Student Result Processing System
#   Tasks:
#     Take input for 5 students (name + marks)
#     Calculate percentage & grade
#     Store results in a .txt file
#     Read file and display toppers
#  Expected Output:
#     Structured file output
#     Sorted results
```

```python
students = []

for i in range(5):
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))
    
    percentage = marks
    if percentage >= 90:
        grade = "A"
    elif percentage >= 75:
        grade = "B"
    elif percentage >= 50:
        grade = "C"
    else:
        grade = "Fail"
    
    students.append((name, percentage, grade))

# Write to file
with open("results.txt", "w") as f:
    for s in students:
        f.write(f"{s[0]} {s[1]} {s[2]}\n")

# Read & display topper
students.sort(key=lambda x: x[1], reverse=True)

print("Topper:", students[0])
```

```python
# Question 2: Build a Data Analyzer Tool
# Tasks:
#     Input a mixed list
#     Separate integers, floats, strings
#     Count frequency of each type
#     Print summary using loops
```

```python
data = [10, 2.5, "hi", 7, "python", 3.2]

ints, floats, strings = [], [], []

for item in data:
    if isinstance(item, int):
        ints.append(item)
    elif isinstance(item, float):
        floats.append(item)
    else:
        strings.append(item)

print("Integers:", ints)
print("Floats:", floats)
print("Strings:", strings)

print("Count -> Int:", len(ints), "Float:", len(floats), "String:", len(strings))
```

```python
# Question 3: Create a Math Utility Module
# Tasks:
#     Factorial using recursion
#     Fibonacci using recursion
#     Lambda for square & cube
    # Compare execution time
```

```python
import time

def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

square = lambda x: x*x
cube = lambda x: x*x*x

start = time.time()
print("Factorial:", factorial(5))
print("Fibonacci:", fibonacci(6))
end = time.time()

print("Time:", end-start)
```

```python
# Question 4: Build an Email Extractor Tool
# Tasks:
#     Read a text file
#     Extract valid emails using regex
#     Count unique emails
#     Save results in CSV
```

```python
import re
import csv

with open("data.txt", "r") as f:
    text = f.read()

emails = re.findall(r'[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.[a-z]{2,}', text)

unique_emails = list(set(emails))

with open("emails.csv", "w", newline="") as f:
    writer = csv.writer(f)
    for email in unique_emails:
        writer.writerow([email])

print("Total unique emails:", len(unique_emails))
```

```python
# Question 6: Design an Employee Management System
# Tasks:
#     Base class: Employee
#     Derived: Manager, Developer
#     Override salary method
#     Display employee details
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def show(self):
        print(self.name, self.salary)

class Manager(Employee):
    def salary_calc(self):
        return self.salary + 5000

class Developer(Employee):
    def salary_calc(self):
        return self.salary + 3000

m = Manager("Aman", 50000)
d = Developer("Riya", 40000)

print("Manager Salary:", m.salary_calc())
print("Developer Salary:", d.salary_calc())
```

```python
# Question 7:  Optimize performance using advanced Python
# Tasks:
#     Create decorator for logging
#     Create generator for large data
#     Compare memory usage
```

```python
def logger(func):
    def wrapper():
        print("Function started")
        func()
        print("Function ended")
    return wrapper

@logger
def display():
    print("Hello")

display()

def generator(n):
    for i in range(n):
        yield i

for num in generator(5):
    print(num)
```

```python
# Question 8: Simulate Parallel File Downloader
# Tasks:
#     Download multiple files using threads
#     Measure execution time
#     Compare with sequential execution
```

```python
# 8
import threading
import time

def download(file):
    print("Downloading", file)
    time.sleep(2)
    print("Completed", file)

files = ["file1", "file2", "file3"]

threads = []

start = time.time()

for f in files:
    t = threading.Thread(target=download, args=(f,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

end = time.time()

print("Time:", end-start)
```

```python
# Question 9: Perform Data Analysis
# Tasks:
#     Load dataset
#     Handle missing values
#     Perform filtering
#     Compute mean, median, aggregation
```

```python
import pandas as pd

data = {
    "Name": ["A", "B", "C"],
    "Marks": [90, None, 75]
}

df = pd.DataFrame(data)

df["Marks"].fillna(df["Marks"].mean(), inplace=True)

print("Mean:", df["Marks"].mean())
print("Filtered:", df[df["Marks"] > 80])
```

```python
# Build a Student Management System (End-to-End)
# Features:
#     Add student
#     Search student
#     Store in CSV
#     Display analytics (topper, avg marks)
```

```python
import csv

def add_student():
    name = input("Enter name: ")
    marks = int(input("Enter marks: "))
    
    with open("students.csv", "a", newline="") as f:
        writer = csv.writer(f)
        writer.writerow([name, marks])

def display():
    with open("students.csv", "r") as f:
        for row in csv.reader(f):
            print(row)

def topper():
    with open("students.csv", "r") as f:
        data = list(csv.reader(f))
        data.sort(key=lambda x: int(x[1]), reverse=True)
        print("Topper:", data[0])

add_student()
display()
topper()
```
