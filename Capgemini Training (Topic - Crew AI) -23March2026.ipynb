```python
# Scenario: AI Content Automation for a Startup
# Problem Statement
# A startup wants to:
# Stay updated with latest AI trends
# Automatically create LinkedIn posts
# Save time on manual research & writing
# So we build a multi-agent AI system to automate this using CrewAI.

import os
from dotenv import load_dotenv
load_dotenv()

from crewai import Agent, Task, Crew
from langchain_litellm import ChatLiteLLM

os.environ["CREWAI_TRACING_ENABLED"] = "true"

llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

researcher = Agent(
    role="Researcher",
    goal="Find latest AI news",
    backstory="Expert in AI trends",
    llm=llm,
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Write LinkedIn post",
    backstory="Expert content creator",
    llm=llm,
    verbose=True
)

research_task = Task(
    description="Find 5 latest AI news trends",
    expected_output="List with summaries",
    agent=researcher
)

write_task = Task(
    description="Write LinkedIn post",
    expected_output="Final post",
    agent=writer,
    context=[research_task]
)

crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    verbose=True
)

result = crew.kickoff()

print(result)
```

```python
# Scenario: Automated Learning Hub for an EdTech Company
# Problem Statement
# An EdTech company wants to:
# - Keep track of the latest educational technology innovations
# - Automatically generate weekly blog posts for their website
# - Reduce time spent on manual research and content drafting
#  So we build a multi-agent AI system using CrewAI to automate this workflow:
# - Trend Monitoring Agent → Continuously scans for new EdTech tools, teaching strategies, and AI-in-education updates
# - Content Summarization Agent → Condenses research into digestible insights tailored for educators and learners
# - Blog Writing Agent → Crafts engaging, SEO-friendly blog posts automatically
# - Publishing Agent → Schedules and posts content directly to the company’s website
import os
from dotenv import load_dotenv
load_dotenv()

from crewai import Agent, Task, Crew
from langchain_litellm import ChatLiteLLM

os.environ["CREWAI_TRACING_ENABLED"] = "true"

llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

trend_monitor = Agent(
    role="Trend Monitoring Agent",
    goal="Find latest EdTech innovations and AI in education",
    backstory="Expert in tracking education technology trends",
    llm=llm,
    verbose=True
)

summarizer = Agent(
    role="Content Summarization Agent",
    goal="Summarize research into simple insights",
    backstory="Expert in simplifying educational content",
    llm=llm,
    verbose=True
)

writer = Agent(
    role="Blog Writing Agent",
    goal="Write SEO-friendly blog posts",
    backstory="Expert blog writer for EdTech",
    llm=llm,
    verbose=True
)

publisher = Agent(
    role="Publishing Agent",
    goal="Prepare content for publishing",
    backstory="Expert in formatting and publishing blogs",
    llm=llm,
    verbose=True
)

trend_task = Task(
    description="Find latest EdTech innovations, AI in education, and new learning tools",
    expected_output="List of trends with brief explanation",
    agent=trend_monitor
)

summary_task = Task(
    description="Summarize trends into easy insights for teachers and students",
    expected_output="Clear summarized insights",
    agent=summarizer,
    context=[trend_task]
)

writing_task = Task(
    description="Write a weekly SEO-friendly blog post on EdTech trends",
    expected_output="Full blog post with title, headings, and content",
    agent=writer,
    context=[summary_task]
)

publishing_task = Task(
    description="Format the blog and prepare it for publishing",
    expected_output="Final ready-to-publish blog content",
    agent=publisher,
    context=[writing_task]
)

crew = Crew(
    agents=[trend_monitor, summarizer, writer, publisher],
    tasks=[trend_task, summary_task, writing_task, publishing_task],
    verbose=True
)

result = crew.kickoff()

print(result)
```

```python
# Scenario: Smart Job Application Assistant
# 🏢 Problem Statement

# Students struggle to:

# Analyze job descriptions

# Match their skills

# Write tailored resumes & cover letters

# 👉 So we build a multi-agent AI system that automates job application preparation.

# 🤖 System Overview
# 🔹 Agent 1: Job Analyzer

# Reads job description

# Extracts required skills

# 🔹 Agent 2: Resume Optimizer

# Matches candidate skills with job

# Suggests improvements

# 🔹 Agent 3: Cover Letter Writer

# Generates a personalized cover letter

# 🔄 Workflow
# Job Description → Skill Analysis → Resume Optimization → Cover Letter Generation
# !pip install crewai litellm langchain-community langchain-litellm python-dotenv

import os
from dotenv import load_dotenv
load_dotenv()

from crewai import Agent, Task, Crew
from langchain_litellm import ChatLiteLLM

os.environ["CREWAI_TRACING_ENABLED"] = "true"

llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

job_analyzer = Agent(
    role="Job Analyzer",
    goal="Extract key skills and requirements from job descriptions",
    backstory="HR expert who analyzes job postings",
    llm=llm,
    verbose=True
)

resume_optimizer = Agent(
    role="Resume Optimizer",
    goal="Align resume with job requirements",
    backstory="Career coach improving resumes",
    llm=llm,
    verbose=True
)

cover_letter_writer = Agent(
    role="Cover Letter Writer",
    goal="Write personalized cover letter",
    backstory="Professional writer for job applications",
    llm=llm,
    verbose=True
)

job_description = """
We are looking for a Data Analyst with experience in Python, SQL, Power BI,
and data visualization. The candidate should be able to analyze datasets,
create dashboards, and communicate insights effectively.
"""

candidate_resume = """
I am a BTech student skilled in Python and basic SQL.
I have created Power BI dashboards and completed projects on data analysis.
"""

analysis_task = Task(
    description=f"""
    Analyze job description and extract:
    - Skills
    - Responsibilities
    - Tools

    {job_description}
    """,
    expected_output="Structured skills and requirements",
    agent=job_analyzer
)

resume_task = Task(
    description=f"""
    Compare resume with job and:
    - Find gaps
    - Suggest improvements
    - Optimize resume

    {candidate_resume}
    """,
    expected_output="Improved resume",
    agent=resume_optimizer,
    context=[analysis_task]
)

cover_letter_task = Task(
    description="""
    Write a professional cover letter using job analysis and improved resume
    """,
    expected_output="Final cover letter",
    agent=cover_letter_writer,
    context=[analysis_task, resume_task]
)

crew = Crew(
    agents=[job_analyzer, resume_optimizer, cover_letter_writer],
    tasks=[analysis_task, resume_task, cover_letter_task],
    verbose=True
)

result = crew.kickoff()

print(result)
```

```python
# ==============================
# Smart Sales Proposal Assistant (Corrected Output Access)
# ==============================

import os
from dotenv import load_dotenv
load_dotenv()

from crewai import Agent, Task, Crew
from langchain_litellm import ChatLiteLLM

os.environ["CREWAI_TRACING_ENABLED"] = "true"

# ==============================
# 1. Groq LLM
# ==============================
llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

# ==============================
# 2. Define Agents
# ==============================
rfp_analyzer = Agent(
    role="RFP Analyzer",
    goal="Extract key requirements, budget, evaluation criteria from RFPs",
    backstory="Expert in reading RFPs and extracting business requirements",
    llm=llm,
    verbose=True
)

solution_mapper = Agent(
    role="Solution Mapper",
    goal="Match company offerings to client requirements and highlight differentiators",
    backstory="Product expert who aligns solutions to client needs",
    llm=llm,
    verbose=True
)

proposal_writer = Agent(
    role="Proposal Writer",
    goal="Generate polished, client-specific proposals ensuring brand alignment",
    backstory="Professional writer for enterprise sales proposals",
    llm=llm,
    verbose=True
)

# ==============================
# 3. Sample RFP & Company Info
# ==============================
rfp_doc = """
Client: XYZ Corp
Budget: $50,000 - $70,000
Requirements:
- Cloud data analytics solution
- Integration with existing ERP system
- Real-time reporting dashboards
- Training for staff
Evaluation Criteria:
- Cost-effectiveness
- Implementation speed
- Product reliability
- Vendor support
"""

company_offerings = """
Our company provides a cloud-based analytics platform with real-time dashboards,
seamless ERP integration, and dedicated training programs. We ensure fast implementation,
24/7 support, and competitive pricing.
"""

# ==============================
# 4. Define Tasks
# ==============================
rfp_task = Task(
    description=f"Analyze the RFP and extract key requirements, budget, and evaluation criteria:\n{rfp_doc}",
    expected_output="Structured RFP insights",
    agent=rfp_analyzer
)

solution_task = Task(
    description=f"Map company offerings to client requirements and suggest tailored value propositions:\n{company_offerings}",
    expected_output="Solution mapping",
    agent=solution_mapper,
    context=[rfp_task]  # Use RFP output
)

proposal_task = Task(
    description="Generate a polished sales proposal using RFP analysis and solution mapping",
    expected_output="Final sales proposal",
    agent=proposal_writer,
    context=[rfp_task, solution_task]  # Use outputs from previous tasks
)

# ==============================
# 5. Initialize Crew
# ==============================
crew = Crew(
    agents=[rfp_analyzer, solution_mapper, proposal_writer],
    tasks=[rfp_task, solution_task, proposal_task],
    verbose=True
)

# ==============================
# 6. Run Workflow
# ==============================
result = crew.kickoff()

# ==============================
# 7. Print Outputs Correctly
# ==============================
print("\n=== RFP ANALYSIS ===")
print(result[rfp_task.name].output if rfp_task.name in result else "No output found")

print("\n=== SOLUTION MAPPING ===")
print(result[solution_task.name].output if solution_task.name in result else "No output found")

print("\n=== SALES PROPOSAL ===")
print(result[proposal_task.name].output if proposal_task.name in result else "No output found")
```

```python
# Enterprise sales teams often struggle to:
# - Analyze client requirements from RFPs (Requests for Proposals)
# - Match company offerings to client needs
# - Draft tailored proposals that stand out
# - Ensure consistency across documents
# 👉 So we build a multi-agent AI system that automates sales proposal preparation.

# 🤖 System Overview
# - 🔹 Agent 1: RFP Analyzer
# - Reads client’s request for proposal
# - Extracts key requirements, budget constraints, and evaluation criteria
# - 🔹 Agent 2: Solution Mapper
# - Matches company products/services to client needs
# - Suggests tailored value propositions and differentiators
# - 🔹 Agent 3: Proposal Writer
# - Generates a polished, client-specific proposal
# - Ensures alignment with brand tone and compliance standards
# 🔄 Workflow
# RFP Document → Requirement Analysis → Solution Mapping → Proposal Generation
import os
from crewai import Agent, Task, Crew, Process
from dotenv import load_dotenv
from langchain_litellm import ChatLiteLLM
load_dotenv()

# -----------------------------
# 1. API Setup
# -----------------------------
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

if "OPENAI_API_KEY" in os.environ:
    del os.environ["OPENAI_API_KEY"]

# -----------------------------
# 2. LLM
# -----------------------------
llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

# -----------------------------
# 3. Agents (More Detailed Roles)
# -----------------------------

rfp_agent = Agent(
    role="Senior RFP Analyst",
    goal="Deeply analyze RFP documents and extract structured business insights",
    backstory="10+ years experience in enterprise sales and bid management",
    llm=llm,
    verbose=True
)

mapping_agent = Agent(
    role="Solution Architect",
    goal="Design tailored solutions matching client requirements",
    backstory="Expert in mapping enterprise solutions and creating winning strategies",
    llm=llm,
    verbose=True
)

proposal_agent = Agent(
    role="Proposal Specialist",
    goal="Create high-quality, persuasive, and structured proposals",
    backstory="Professional writer specializing in enterprise proposals and documentation",
    llm=llm,
    verbose=True
)

# -----------------------------
# 4. Input Data
# -----------------------------
rfp_text = """
Client needs a scalable AI-powered analytics platform with dashboarding,
real-time insights, and API integrations. Budget is flexible but prefers cost-efficient solutions.

Evaluation based on:
- Innovation
- Scalability
- Pricing
"""

company_data = """
We provide AI-driven analytics solutions with real-time dashboards,
cloud scalability, and API integrations. Our platform is cost-effective and easy to use.
"""

# -----------------------------
# 5. Tasks (More Structured Prompts)
# -----------------------------

task1 = Task(
    description=f"""
    Extract structured insights from the RFP.

    Format:
    {{
      "requirements": [],
      "budget": "",
      "evaluation_criteria": [],
      "objectives": []
    }}

    RFP:
    {rfp_text}
    """,
    expected_output="JSON structured analysis",
    agent=rfp_agent
)

task2 = Task(
    description=f"""
    Using the RFP analysis and company data:

    - Map each requirement to a solution
    - Provide justification
    - Highlight competitive advantages

    Return in structured format.
    
    Company Info:
    {company_data}
    """,
    expected_output="Structured mapping with justification",
    agent=mapping_agent,
    context=[task1]
)

task3 = Task(
    description="""
    Generate a complete enterprise proposal document.

    Structure:
    1. Executive Summary
    2. Client Requirements Understanding
    3. Proposed Solution
    4. Key Benefits
    5. Why Choose Us
    6. Conclusion

    Tone: Professional, persuasive, and business-oriented
    """,
    expected_output="Full proposal document",
    agent=proposal_agent,
    context=[task1, task2]
)

# -----------------------------
# 6. Crew (Sequential Process)
# -----------------------------
crew = Crew(
    agents=[rfp_agent, mapping_agent, proposal_agent],
    tasks=[task1, task2, task3],
    process=Process.sequential,   # Important difference
    verbose=True
)

# -----------------------------
# 7. Run
# -----------------------------
output = crew.kickoff()

print("\n🚀 FINAL PROPOSAL:\n")
print(output)
```
